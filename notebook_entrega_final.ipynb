# ENTREGA FINAL – Dashboard de Control de Obra Vial

Base de datos sintética generada con Python y apoyo de IA. El objetivo es analizar producción, cumplimiento, costos, equipos, incidencias y calidad.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('base_obra_vial_ia.csv')
df['fecha'] = pd.to_datetime(df['fecha'])
display(df.head())
print('Registros:', len(df))
```

## 1. Validación de la base


```python
print(df.info())
print('\nValores nulos:')
print(df.isna().sum())
```

## 2. Indicadores principales


```python
prog = df['produccion_programada'].sum()
real = df['produccion_real'].sum()
cumplimiento = real / prog * 100
costo_prog = df['costo_programado_soles'].sum()
costo_real = df['costo_real_soles'].sum()

print('Producción programada:', round(prog, 2))
print('Producción real:', round(real, 2))
print('Cumplimiento (%):', round(cumplimiento, 2))
print('Costo programado (S/):', round(costo_prog, 2))
print('Costo real (S/):', round(costo_real, 2))
print('Horas de parada:', round(df['horas_parada'].sum(), 2))
```

## 3. Cumplimiento promedio por partida


```python
rend = df.groupby('partida')['cumplimiento_pct'].mean().sort_values()
display(rend)
rend.plot(kind='barh', title='Cumplimiento promedio por partida')
plt.xlabel('%')
plt.tight_layout()
plt.show()
```

## 4. Conclusión

La información sintética permite construir un tablero de control para observar avance, rendimiento, costos e incidencias de una obra vial. El dashboard interactivo final se encuentra en `app.py` y se publica mediante Streamlit.
